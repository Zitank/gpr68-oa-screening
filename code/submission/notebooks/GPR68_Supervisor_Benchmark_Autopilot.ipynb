# GPR68 Supervisor Benchmark Autopilot

**Supervisor design implemented exactly:** Pocket 1 is the primary candidate; Pocket 2 is the alternative site. The same ligand set is docked into both sites with identical GNINA settings except for the pocket centre.

This notebook runs the benchmark only — no 50-compound pilot and no full-library/DDL step.


```python
from google.colab import files
import zipfile, os, glob, shutil
print('Upload GPR68_Supervisor_Benchmark_Autopilot.zip')
u = files.upload()
zips = [x for x in u if x.lower().endswith('.zip')]
assert zips, 'No ZIP uploaded.'
z = zips[0]
base = '/content/GPR68_SUPERVISOR_RUN'
shutil.rmtree(base, ignore_errors=True)
os.makedirs(base, exist_ok=True)
with zipfile.ZipFile(z) as f:
    f.extractall(base)
scripts = glob.glob(base + '/**/autopilot.py', recursive=True)
assert scripts, 'autopilot.py not found in uploaded ZIP.'
SCRIPT = scripts[0]
print('Autopilot script:', SCRIPT)
```
